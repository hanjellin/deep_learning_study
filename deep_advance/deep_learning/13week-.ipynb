```python
import nltk
import re
import urllib.request
import zipfile
from lxml import etree
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
nltk.download('all')
```

```python
targetXML = open('ted_en-20160408.xml','r', encoding= 'UTF8')
target_text = etree.parse(targetXML)
```

```python
parse_text = '\n'.join(target_text.xpath('//content/text()'))
```

```python
content_text = re.sub(r'\([^)]*\)', '', parse_text)
```

```python
sent_text = sent_tokenize(content_text)
```

```python
normalized_text = []
for string in sent_text:
    tokens = re.sub(r"[^a-z0-9]+", " ", string.lower())
    normalized_text.append(tokens)
```

```python
from gensim.models import Word2Vec
from gensim.models import KeyedVectors

model = Word2Vec(sentence = result, vector_size = 100, window = 5, min_count = 4, workers = 4, sg= 0)
```

```python
model_result = model.wv.most_similar("man")
print(model_result)
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np

word_vectors = model.wv
all_vocabs = word_vectors.key_to_index.keys()
vocabs = list(all_vocabs)[:400]
word_vectors_list = [word_vectors[v] for v in vocabs]
word_vectors_array = np.array(word_vectors_list)

mode12 = TSNE(learning_rate = 100)
transformd = mode12.fit_transform(word_vectors_array)
xs2 = transformd[:, 0]
ys2 = transformd[:, 1]

plt.figure(figsize=(28,21))
plt.scatter(xs2, ys2)

for i ,v in enumerate(vocabs):
    plt.annotate(v,xy=(xs2[i], ys2[i]))

plt.title("Word Vector Visualization using t-SNE")
plt.show()
```
